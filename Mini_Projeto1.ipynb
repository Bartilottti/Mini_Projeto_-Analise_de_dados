```python
from ast import Return
from numpy.random.mtrand import randint
def gera_dados_ficticios(num_registros = 600):

  print(f"\nIniciando a geração de {num_registros} registros de vendas...\n")

  produtos = {
    'Notebook Gamer' : {'categorias' : 'computadores', 'preço' : 1200.00},
    'Ryzen 7 5700g' : {'categorias' : 'processadores', 'preço' : 850.00},
    'RTX 5090' : {'categorias' : 'Placas de video', 'preço' : 4500.00},
    'SSD 1TB' : {'categorias' : 'Armazenamento', 'preço' : 700.00},
    'Gabinete aquario' : {'categorias' : 'Perifericos', 'preço' : 300.00},
    'Mouse atackshark' : {'categorias' : 'Perifericos', 'preço' : 120.00},
    'Teclado Machinike B500' : {'categorias': 'Perifericos', 'preço' : 180.00},
    'Cadeira Gamer' : {'categorias' : 'Moveis', 'preço' : 450.00},
    'Pasta termica PCYES' : {'categorias' : 'Manutenção', 'preço' : 100.00},
    'Headset' : {'categorias' : 'SOM', 'preço' : 190.00}
}

  lista_produtos = list(produtos.keys())

  cidades_estado = {
    'Salvador' : 'BA', 'Aracaju' : 'SE', 'Recife' : 'PE' , 'Fortaleza' : 'CE',
    'Maceió' : 'AL', 'Natal' : 'RGN', 'São Luis' : 'MA'
  }

  Lista_cidades = list(cidades_estado.keys())

  dados_vendas = []

  data_inicial = datetime(2026,1,1)

  for i in range(num_registros):
    produto_nome = random.choice(lista_produtos)

    cidade_nome = random.choice(Lista_cidades)

    quantidade = np.random.randint(1,8)

    data_pedido = data_inicial + timedelta(days = int(i/5), hours = random.randint(0,23))

    if produto_nome in['Mouse atackshark', 'Teclado Machinike B500', 'Headset']:
      preco_unitario = produtos[produto_nome]['preço'] * np.random.uniform(0.9,1.0)
    else:
      preco_unitario = produtos[produto_nome]['preço']

    dados_vendas.append({
      'ID_pedido' : 1000 + i,
      'Data Pedido' : data_pedido,
      'Nome_Produto' : produto_nome,
      'Categoria' : produtos[produto_nome]['categorias'],
      'preço Unitario' : round(preco_unitario, 2),
      'Quantidade' : quantidade,
      'ID Cliente' : np.random.randint(100,150),
      'Cidade' : cidade_nome,
      'Estado' : cidades_estado[cidade_nome]
    })

  print('Geração dos dados foi concluida')

  return pd.DataFrame(dados_vendas)
```

```python
df_vendas = gera_dados_ficticios(500)
```

```python
df_vendas['Data Pedido'] = pd.to_datetime(df_vendas['Data Pedido'])
```

```python
df_vendas['Faturamento'] = df_vendas['preço Unitario'] * df_vendas['Quantidade']
```

```python
df_vendas['Tipo de Entrega'] = df_vendas['Estado'].apply(lambda estado: 'Rápida' if estado in ['BA', 'SE', 'PE'] else 'Normal')
```

```python
df_vendas.info()
```

```python
top_10_produtos = df_vendas.groupby('Nome_Produto')['Quantidade'].sum().sort_values(ascending = False).head(10)
```

```python
top_5_faturamentos = df_vendas.groupby(['Cidade', 'Estado'])['Faturamento'].sum().nlargest(5).reset_index()
```

```python
sns.set_style('whitegrid')

plt.figure(figsize = (12,7))

top_10_produtos.sort_values(ascending = True).plot(kind = 'barh', color = 'skyblue')

plt.title('Top 10 Produtos mais vendidos', fontsize = 16)
plt.xlabel('Quantidade Vendida', fontsize = 12)
plt.ylabel('Produto', fontsize = 12)

plt.tight_layout()
plt.show()
```

```python
top_5_faturamentos['Local'] = top_5_faturamentos['Cidade'] + ' - ' + top_5_faturamentos['Estado']

plt.figure(figsize=(10, 6))

plt.bar(top_5_faturamentos['Local'], top_5_faturamentos['Faturamento'], color='royalblue')

plt.title('Top 5 Faturamentos por Cidade')
plt.xlabel('Localidade')
plt.ylabel('Faturamento')

plt.xticks(rotation=45)


plt.tight_layout()


plt.show()
```

```python
df_vendas['Mes'] = df_vendas['Data Pedido'].dt.to_period('M')
```

```python
faturamento_mensal = df_vendas.groupby('Mes')['Faturamento'].sum()
```

```python
faturamento_mensal.index = faturamento_mensal.index.strftime('%Y-%m')
```

```python
faturamento_mensal.map('R$ {:,.2f}'.format)
```

```python

plt.figure(figsize=(10, 5))
plt.plot(faturamento_mensal.index, faturamento_mensal.values, marker='o', color='royalblue', linewidth=2)


plt.title('Faturamento Mensal', fontsize=14)
plt.xlabel('Mês', fontsize=12)
plt.ylabel('Faturamento (R$)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```

```python
vendas_estados = df_vendas.groupby('Estado')['Faturamento'].sum().sort_values(ascending=False)
```

```python
vendas_estados.map('R${:,.2f}'.format)
```

```python


plt.figure(figsize=(12,7))

vendas_estados.plot(kind='bar',color=sns.color_palette('husl',7))

plt.title('Faturamento por estado', fontsize=16)

plt.xlabel('Estados',fontsize=12)

plt.ylabel('Faturamento(R$)', fontsize=12)

plt.xticks(rotation=0)

plt.tight_layout()

plt.show()
```

```python
faturamento_categoria = df_vendas.groupby('Categoria')['Faturamento'].sum().sort_values(ascending=False)
```

```python
faturamento_categoria.map('R${:,.2f}'.format)
```

```python
from matplotlib.ticker import FuncFormatter

faturamento_ordenado = faturamento_categoria.sort_values(ascending=False)

fig, ax = plt.subplots(figsize = (12,7))

def formatador_milhares(y,pos):
  return f'R$ {y/1000:,.0f}K'


formatter = FuncFormatter(formatador_milhares)

ax.yaxis.set_major_formatter(formatter)

faturamento_ordenado.plot(kind='bar', ax = ax, color = sns.color_palette('viridis', len(faturamento_ordenado)))

plt.title('Faturamento por Categoria', fontsize=16)
plt.xlabel('Categorias', fontsize=12)
plt.ylabel('Faturamento(R$)')

plt.xticks(rotation = 45, ha = 'right')

plt.tight_layout()
plt.show()
```
